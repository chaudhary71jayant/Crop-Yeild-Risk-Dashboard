# Production Data Cleaning and Historical Dataset

## Crop Yield Risk & Advisory Dashboard

### This notebook loads district-level Area, Production and Yield data for multiple years from the Government of India UPAg portal, standardizes the datasets, filters the project scope, and creates a historical production dataset.

```python
import pandas as pd
```

```python
# Loading the datset from 2021 to 2025
df_2021 = pd.read_csv(
    "../../data/raw/DES-District-Data-For-2021-22.csv"
)

df_2022 = pd.read_csv(
    "../../data/raw/DES-District-Data-For-2022-23.csv"
)

df_2023 = pd.read_csv(
    "../../data/raw/DES-District-Data-For-2023-24.csv"
)

df_2024 = pd.read_csv(
    "../../data/raw/DES-District-Data-For-2024-25.csv"
)
```

```python
target_states = [
    "Uttar Pradesh",
    "Punjab",
    "Haryana"
]
```

```python
# Checking for target state in every year 

for year , dataset in [ ("2021-22", df_2021),
    ("2022-23", df_2022),
    ("2023-24", df_2023),
    ("2024-25", df_2024)
]:
    print("\nYEAR:", year)

    for state in target_states:
        print(state , (dataset["State"]==state).sum())
```

```python
# Checking for target crop in every year

target_crops = [
    "Wheat",
    "Rice",
    "Maize"
]

```

```python
for year , dataset in [ ("2021-22", df_2021),
    ("2022-23", df_2022),
    ("2023-24", df_2023),
    ("2024-25", df_2024)
]:
    print("\nYEAR:", year)

    for crop in target_crops:
        print(crop , (dataset["Crop"]==crop).sum())
```

```python
# Creating a function to processed the data of every year 

def clean_production_data(df , year):

    filtered_df = df[
        df["State"].isin(target_states)&
        df["Crop"].isin(target_crops)&
        (df["Season"]=="Total")
        ].copy()

    area_column = f"Area-{year}"
    production_column = f"Production-{year}"
    yield_column = f"Yield-{year}"

    filtered_df = filtered_df.rename(
        columns={
            area_column : "Area",
            production_column : "Production",
            yield_column : "Yield"}
    )

    filtered_df["Year"] = year

    filtered_df = filtered_df[
        [
            "State",
            "District",
            "Crop",
            "Year",
            "Area",
            "Production",
            "Yield"
        ]
    ]

    return filtered_df

    
```

```python
# Calling the function 

clean_2021 = clean_production_data(
    df_2021,
    "2021-22"
)

clean_2022 = clean_production_data(
    df_2022,
    "2022-23"
)

clean_2023 = clean_production_data(
    df_2023,
    "2023-24"
)

clean_2024 = clean_production_data(
    df_2024,
    "2024-25"
)
```

```python
# Concatinating the production_data

production_history = pd.concat(
    [
        clean_2021,
        clean_2022,
        clean_2023,
        clean_2024
    ],
    ignore_index = True )
```

```python
# Checking for any whitespaces in text values

text_columns = ["State" , "District" , "Crop"]

for column in text_columns:
    production_history[column]=(
        production_history[column].astype("string").str.strip().str.replace(r"\s+"," " , regex = True))
```

```python
# Adding an ordinary variable for year 

year_order = {
    "2021-22": 2021,
    "2022-23": 2022,
    "2023-24": 2023,
    "2024-25": 2024
}

production_history["Year_Order"] = (
    production_history["Year"].map(year_order)
)
```

```python
# Sorting the historical data 

production_history = production_history.sort_values(
    by = ["State" , "District" , "Crop" , "Year_Order"]).reset_index(drop=True)
```

```python
# Chekcing if district have all same years or not 

history_coverage = (
    production_history.groupby(["State" , "District" , "Crop"])["Year"].nunique().reset_index(name = "Years_ Available"))
    
```

```python
history_coverage.head()
```

```python
history_coverage["Years_ Available"].value_counts().sort_index()
```

```python
# Creating the missing years district/Crops flag

history_coverage["Complete_4_year_history"]=(history_coverage["Years_ Available"]==4)
```

```python
history_coverage["Complete_4_year_history"].value_counts()
```

```python
# Attaching this information to the main dataset

production_history = production_history.merge(
    history_coverage ,
    on = ["State" , "District" , "Crop"],
    how = "left"
    )
    
```

```python
production_history.head()
```

```python
# Inspecting one district 

example = production_history[
    (production_history["State"] == "Uttar Pradesh")
    & (production_history["District"] == "Meerut")
    & (production_history["Crop"] == "Wheat")
]


example
```

```python
# Each record in dataset must know previous year_yield

group_columns =[
    "State","District" , "Crop"]

production_history["Previous_Yield"]=(
    production_history.groupby(group_columns)["Yield"].shift(1))
    
```

```python
production_history.head()
```

```python
# Calculating yield change

previous_yield_safe = production_history["Previous_Yield"].replace(0 , pd.NA)

production_history["Previous_yield_percentage"]=(
    (
        production_history["Yield"]-production_history["Previous_Yield"])/ previous_yield_safe)*100
```

```python
# Caluclate Area Percentage 

production_history["Previous_area"]=(
    production_history.groupby(group_columns)["Area"].shift(1))
```

```python
previous_area_safe = (
    production_history["Previous_area"].replace(0 , pd.NA))

production_history["Previous_Area_Percentage"] = (
    (
        production_history["Area"]-production_history["Previous_area"])/previous_area_safe)*100
```

```python
# Calculating production change percentage

production_history["Previous_Production"] = (
    production_history.groupby(group_columns)["Production"].shift(1))

previous_production_safe=(
    production_history["Previous_Production"].replace(0 , pd.NA))

production_history["Production_Change_Pct"] = (
    (
        production_history["Production"]
        - production_history["Previous_Production"]
    )
    / previous_production_safe
) * 100
    
```

```python
# History Average Yiled 

production_history["History_avg_yield"]=(
    production_history.groupby(group_columns)["Yield"].transform(
        lambda x: x.shift(1).expanding().mean()
    )
)
```

```python
# Calculating Yield deviation from historical average

historical_yield_safe = (
    production_history["History_avg_yield"].replace(0 , pd.NA))

production_history["Yield_Deviation_Pct"] = (
    (
        production_history["Yield"]
        - production_history["History_avg_yield"]
    )
    / historical_yield_safe
) * 100
```
