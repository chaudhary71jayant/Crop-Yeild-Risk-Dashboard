## Building rainfall feature correctly


```python
# Sort rainfall chronologically

year_order = {
    "2021-22": 2021,
    "2022-23": 2022,
    "2023-24": 2023,
    "2024-25": 2024
}

rainfall["Year_Order"] = rainfall["Year"].map(
    year_order
)

rainfall = rainfall.sort_values(
    [
        "State",
        "District",
        "Year_Order"
    ]
).reset_index(drop=True)
```

```python
# Calculate historical average rainfall

rainfall[
    "Historical_Avg_Rainfall"
] = (
    rainfall
    .groupby(
        ["State", "District"]
    )["Crop_Year_Rainfall_mm"]
    .transform(
        lambda x:
        x.shift(1).expanding().mean()
    )
)
```

```python
# Calculate rainfall deviation

import numpy as np

historical_rainfall_safe = (
    rainfall[
        "Historical_Avg_Rainfall"
    ]
    .replace(0, np.nan)
)
```

```python
rainfall[
    "Rainfall_Deviation_Pct"
] = (
    (
        rainfall[
            "Crop_Year_Rainfall_mm"
        ]
        -
        rainfall[
            "Historical_Avg_Rainfall"
        ]
    )
    /
    historical_rainfall_safe
) * 100
```

```python
# Checking the new columns

rainfall[
    [
        "State",
        "District",
        "Year",
        "Crop_Year_Rainfall_mm",
        "Historical_Avg_Rainfall",
        "Rainfall_Deviation_Pct"
    ]
].head(15)
```

```python
# Removing helper column

rainfall = rainfall.drop(
    columns=["Year_Order"]
)
```

```python
# Save the corrected rainfall feature dataset
rainfall.to_csv(
    "../../data/processed/"
    "rainfall_features_2021_2025.csv",
    index=False
)
```

```python
# Recreating the merge

agri_data = production.merge(
    rainfall,
    on=[
        "State",
        "District",
        "Year"
    ],
    how="left",
    validate="many_to_one"
)
```
